```python
import pandas as pd
file_path = 'vehicles.csv'
try:
    data = pd.read_csv(file_path)
    print("Dataset successfully loaded!")
   # print(data.head())
    print(data.tail())
    print(data.shape)# it returns the shape if my dataframe like the total number of rows and columns

    for col in data.columns:
        print(data[col].value_counts()) # all the columns 

    print(data["color"].value_counts)# on specific columns values

    print(data.columns.tolist())# to return the names of columns

    print(data.dtypes)# returns the data types of our columns all

    print(data.nunique()) #counting unique categories in the columuns()

    categorical_columns= data.select_dtypes(include=["object"]).columns
    for col in categorical_columns:
        unique_values = data[col].unique()
        print(f"{col}: {len(unique_values)}")
 
    print("Checking null values")
    print(data.isnull().sum())

    """ errors parameters can contain{
    raise: (default)raises an error for invalid parsing.
    coerce: converts invalid parsing to NaN
    ignore: return the original data without any change
    }
    """


except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found.")
except pd.errors.EmptyDataError:
    print("Error: The file is empty.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```

```python
import matplotlib.pyplot as plt # a plotting library
import seaborn as sns #a data visualization library based on Matplotlib
from matplotlib import style #for appearance of your plots by applying different styles


sns.set(rc={ #rc stands for "runtime configuration".  This sets the default figure size for Seaborn plots to 20x20 inches.
    'figure.figsize': (20, 20),        # Set figure size to 10x6 inches
    'font.size': 14,                   # Set default font size to 14
    'axes.titlesize': 16,              # Set axes title font size to 16
    'xtick.labelsize': 12,             # Set x-tick label font size to 12
    'ytick.labelsize': 12,             # Set y-tick label font size to 12
    'lines.linewidth': 2,              # Set default line width to 2
    'axes.labelsize': 14               # Set axes label font size to 14
})
%matplotlib inline

# A stacked KDE plot is a type of data visualization that combines multiple Kernel Density Estimate (KDE) plots into a single graph,
# where each KDE is stacked on top of the others.
# This allows for easy comparison of the distributions of different groups within a dataset.
g = sns.FacetGrid(data, hue='transmission', height=5, aspect=2, xlim=(data["proposed_purchase_price"].min(),data["proposed_purchase_price"].max())) #
# Map the KDE plot
g.map(sns.kdeplot, 'proposed_purchase_price', fill=True)
g.add_legend(title='Purchase Price')
g.set_axis_labels('Purchase Price', 'Density')
g.fig.suptitle('KDE of Purchase Price by transmission', fontsize=16)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loop through the specified columns
for col in ['manufacturer', 'year', 'color', 'body_type', 'engine_type', 'transmission', 
            'fuel_type', 'seating_capacity', 'vehicle_condition_status', 'owner_profession', 'owner_age']:
    try:
        plt.figure(figsize=(23, 5))
        
        # Create the countplot
        ax = sns.countplot(x=col, data=data, palette='colorblind')
        
        # Add grid and title
        plt.grid(color="green", linestyle="--", linewidth=0.5)
        plt.title(f"{col.capitalize()} with their counts", fontdict={"family": "serif", "color": "blue", "size": 20})
        plt.xticks(rotation=90)  # Rotate x labels for better visibility
        
        # Add counts on top of each bar
        for p in ax.patches:
            ax.annotate(f'{int(p.get_height())}', 
                        (p.get_x() + p.get_width() / 2., p.get_height()), 
                        ha='center', va='bottom', fontsize=12, color='green', rotation=90)
        
        # Show plot
        plt.show()
    except Exception as e:
        print(f"An error occurred while processing the column '{col}': {e}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Define the column to plot
column_to_plot = 'manufacturer'

# Create the count plot
plt.figure(figsize=(12, 6))  # Adjust figure size as needed
ax = sns.countplot(x=column_to_plot, data=data, palette='colorblind')

# Add grid and title
plt.grid(color="green", linestyle="--", linewidth=0.5)
plt.title(f"{column_to_plot.capitalize()} Count Plot", fontdict={"family": "serif", "color": "blue", "size": 16})
plt.xlabel(column_to_plot.capitalize(), fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.xticks(rotation=45)  # Rotate x-axis labels for better visibility

# Annotate counts on top of bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='bottom', fontsize=10, color='green')

# Show the plot
plt.tight_layout()  # Adjust layout to fit elements
plt.show()
```

```python
for col in [
    'manufacturer',
    'year',
    'color',
    'body_type',
    'engine_type',
    'transmission',
    'fuel_type',
    'seating_capacity',
    'vehicle_condition_status',
    'owner_profession',
    'owner_age']:
    plt.figure(figsize=(23, 5))
    # Count values for the current column
    counts = data[col].value_counts().sort_index()
    # Create a line chart
    plt.plot(counts.index, counts.values, marker='o', linestyle='-', color='blue')
    # Customize the plot
    plt.grid(color="green", linestyle="--", linewidth=0.5)
    plt.title(f"{col.capitalize()} Counts", fontdict={"family": "serif", "color": "blue", "size": 20})
    plt.xlabel(col.capitalize(), fontsize=14)
    plt.ylabel('Counts', fontsize=14)
    plt.xticks(rotation=90)  # Rotate x labels for better visibility
    # Annotate counts on the line chart
    for x, y in zip(counts.index, counts.values):
        plt.text(x, y, str(y), ha='center', va='bottom', fontsize=12, color='green')
    plt.tight_layout()  # Adjust layout
    plt.show()
```

```python
#Histograms for outliers (selling price)
plt.figure(figsize=(15,3))
sns.histplot(data["selling_price"],color="blue", bins=[2000,11000,25000,34000])
plt.title("Distribution of Selling Price (Including Outliers)", fontsize=14)
ax = plt.gca()
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        plt.text(
            p.get_x() + p.get_width() / 2,
            height,
            f"{int(height)}",
            ha="center",
            va="bottom",
            fontsize=10
        )
plt.xlabel("Selling Price")
plt.ylabel("Frequency")
plt.show()
```
